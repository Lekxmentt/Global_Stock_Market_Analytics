# Step 4: Read CSV files and write them to delta tables

```python

for file in INGESTION_FILES:
    
    # Reading csv file
    print(f"Reading file: {file["file_name"]}")
    df = spark.read.csv(file["file_path"], header=True, inferSchema=True)

    # minor data pre-processing 
    if "Adj Close" in df.columns:
        df = df.withColumnRenamed("Adj Close", "Adj_Close")

    # Writing to delta table
    print(f"Writing to table: {file["file_name"]}")
    df.write.format("delta").mode("overwrite").saveAsTable(file["file_name"])

    print("Done!")
    print()

print("Success !!")
```
